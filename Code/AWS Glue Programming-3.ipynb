```python_glue_session
%idle_timeout 2880
%glue_version 4.0
%worker_type G.1X
%number_of_workers 2

import sys
from awsglue.transforms import *
from awsglue.utils import getResolvedOptions
from pyspark.context import SparkContext
from awsglue.context import GlueContext
from awsglue.job import Job
  
sc = SparkContext.getOrCreate()
glueContext = GlueContext(sc)
spark = glueContext.spark_session
job = Job(glueContext)
```

#### Example: Create a dataframe from a table in the AWS Glue Data Catalog & print  records count in Dataframe


```python_glue_session
dyf = glueContext.create_dynamic_frame.from_catalog(database='bt-course-db-final', table_name='raw_data_2024_mm_06')
df = dyf.toDF()
print("Total records count in dataframe: "+str(df.count()))
df.show(120,truncate=False)
```

#### Example: Get Distinct Rows (By Comparing All Columns)

```python_glue_session
distinctDF = df.distinct()
print("Distinct count: "+str(distinctDF.count()))
distinctDF.show(truncate=False)
```

#### Example: Drops duplicate rows using dropDuplicates() function (By Comparing All Columns)

```python_glue_session
# Applying dropDuplicates() to remove duplicates
df2 = df.dropDuplicates()
print("Distinct count: "+str(df2.count()))
df2.show(truncate=False)
```

#### Example : Drops duplicate rows using dropDuplicates() function (By Comparing selected columns)

```python_glue_session
# Applying dropDuplicates() to remove duplicates
#df2 = df.dropDuplicates(["date","time"])
#print("Distinct count: "+str(df2.count()))
#df2.show(truncate=False)

# Applying dropDuplicates() to remove duplicates
df2 = df.dropDuplicates(["euipment id","equipment name"])
print("Distinct count: "+str(df2.count()))
df2.show(truncate=False)
```

#### Question : How is distinct() different from dropDuplicates()?

##### Answer : Distinct() and dropDuplicates() in PySpark are used to remove duplicate rows, but there is a subtle difference. distinct() considers all columns when identifying duplicates, while dropDuplicates() allowing you to specify a subset of columns to determine uniqueness.

##### Answer: In pyspark, both orderBy() and sort() are methods used for sorting rows in a DataFrame, and they serve the same purpose.
